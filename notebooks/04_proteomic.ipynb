```python
import pandas as pd

df = pd.read_csv("../data/raw/proteomic/proteinGroups.txt", sep="\t")

# remove contaminants
df = df[df["Reverse"] != "+"]
df = df[df["Potential contaminant"] != "+"]

# select reporter intensity columns
cols = [c for c in df.columns if "Reporter intensity corrected" in c]

expr = df[cols]

# set protein names
expr.index = df["Majority protein IDs"]

# transpose → samples x proteins
X = expr.T
print(X)
```
