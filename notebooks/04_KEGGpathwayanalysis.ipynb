```python
import pandas as pd
import numpy as np
import gseapy as gp


```

```python


genesets = gp.get_library(
    name="KEGG_2021_Human"
)

reactome = gp.get_library(
    name="Reactome_2022"
)


apoptosis = genesets["Apoptosis"]
ferroptosis = genesets["Ferroptosis"]
necroptosis = genesets["Necroptosis"]
autophagy = genesets["Autophagy"]
pyroptosis = reactome["Pyroptosis R-HSA-5620971"]

cell_death_genes = {
    "Apoptosis": set(apoptosis),
    "Ferroptosis": set(ferroptosis),
    "Necroptosis": set(necroptosis),
    "Pyroptosis": set(pyroptosis),
    "Autophagy": set(autophagy)
}
```

```python
subtypes_names = ["idh","classical","mesenchymal","proneural"]

subtypes = {}

for name in subtypes_names:
    subtypes[name] = pd.read_csv(f"../data/processed/diff/{name}_combined.csv",index_col=0)

cell_death_hits = {}

for subtype, results in subtypes.items():

    cell_death_hits[subtype] = {}

    for death_type, genes in cell_death_genes.items():

        hits = results.loc[
            results.index.intersection(genes)
        ]

        hits = hits[
            hits["adjustedpvalue_rna"] < 0.05
        ]

        cell_death_hits[subtype][death_type] = hits
        
for name in subtypes_names:
    for type in cell_death_genes.keys():
        cell_death_hits[name][type].to_csv(f"../data/processed/pathway_diff/{name}/{name}_{type}_diff.csv")
```
